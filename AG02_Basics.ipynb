## 1. What is AutoGen?

AutoGen is a framework for building applications with multiple AI agents.

An **agent** is like a role-based worker.  
For example:

- A planner agent can break a task into steps.
- A coder agent can write code.
- A reviewer agent can check the answer.
- A user proxy agent can represent human input.

AutoGen helps these agents communicate and complete tasks together.

## 2. Basic Agent Idea

Before using the real AutoGen library, let us understand the idea with a simple Python class.

````python
class SimpleAgent:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def respond(self, message):
        return f"{self.name} ({self.role}) received: {message}"


agent = SimpleAgent("TrainerAgent", "Explains GenAI concepts clearly")
print(agent.respond("What is AutoGen?"))
````

## 3. Creating Multiple Agents

In real AutoGen projects, we usually create multiple agents with different responsibilities.

````python
planner = SimpleAgent("PlannerAgent", "Breaks the problem into steps")
coder = SimpleAgent("CoderAgent", "Writes Python code")
reviewer = SimpleAgent("ReviewerAgent", "Reviews and improves the answer")

task = "Create a Python function to add two numbers"

print(planner.respond(task))
print(coder.respond(task))
print(reviewer.respond(task))
````

## 4. Simulating Agent Collaboration

Here we simulate how agents pass information from one to another.

````python
task = "Create a Python function to calculate square of a number"

planner_output = "Step 1: Define function. Step 2: Take input. Step 3: Return number * number."
coder_output = '''
def square_number(n):
    return n * n
'''
reviewer_output = "The code is simple and correct. Add type hints for better quality."

print("TASK:")
print(task)

print("\nPLANNER OUTPUT:")
print(planner_output)

print("\nCODER OUTPUT:")
print(coder_output)

print("\nREVIEWER OUTPUT:")
print(reviewer_output)
````

## 5. Why Multi-Agent Systems?

A single LLM call can answer a question, but multi-agent systems are useful when the task needs:

- Planning
- Coding
- Review
- Tool usage
- Human approval
- Iteration
- Role-based reasoning

Example:  
For test automation, one agent can read requirements, another can generate test cases, and another can review edge cases.

|---|---|

## 9. First Real AutoGen Agent Example

This example needs an OpenAI API key.

Set your key as environment variable:

Windows PowerShell:

```powershell
$env:OPENAI_API_KEY="your_api_key_here"
```

Mac/Linux:

```bash
export OPENAI_API_KEY="your_api_key_here"
```

````python
import os
import asyncio

async def run_first_autogen_agent():
    try:
        from autogen_agentchat.agents import AssistantAgent
        from autogen_ext.models.openai import OpenAIChatCompletionClient
    except ImportError:
        print("AutoGen is not installed. Please install it first.")
        return

    if not os.getenv("OPENAI_API_KEY"):
        print("OPENAI_API_KEY not found. Skipping live AutoGen call.")
        return

    model_client = OpenAIChatCompletionClient(
        model="gpt-4o-mini",
        api_key=os.getenv("OPENAI_API_KEY")
    )

    assistant = AssistantAgent(
        name="genai_trainer",
        model_client=model_client,
        system_message="You are a helpful GenAI trainer. Explain concepts clearly with examples."
    )

    response = await assistant.run(task="Explain AutoGen in simple words with one example.")
    print(response)

# In Jupyter, run:
# await run_first_autogen_agent()
````

## 11. Tool Calling Concept

Agents become more powerful when they can use tools.

A tool is a normal Python function that an agent can call.

Example tools:
- calculator
- database query
- API call
- file reader
- test case generator

````python
def add_numbers(a: int, b: int) -> int:
    return a + b

def multiply_numbers(a: int, b: int) -> int:
    return a * b

print("Addition:", add_numbers(10, 20))
print("Multiplication:", multiply_numbers(5, 6))
````

## 12. Mock Tool-Using Agent

This example does not need any API key.

````python
class ToolUsingAgent:
    def __init__(self, name):
        self.name = name

    def solve(self, operation, a, b):
        if operation == "add":
            return add_numbers(a, b)
        elif operation == "multiply":
            return multiply_numbers(a, b)
        else:
            return "Unsupported operation"


tool_agent = ToolUsingAgent("CalculatorAgent")

print(tool_agent.solve("add", 15, 25))
print(tool_agent.solve("multiply", 7, 8))
````

## 13. AutoGen Use Case for Test Automation

AutoGen can be useful in QA/SDET work.

Example workflow:

1. Requirement Agent reads user story.
2. Test Case Agent creates positive and negative test cases.
3. Automation Agent creates Selenium/WebDriverIO code.
4. Review Agent checks missing edge cases.
5. Human approves final output.

````python
requirement = '''
As a user, I should be able to login using valid username and password.
If credentials are invalid, I should see an error message.
'''

test_cases = [
    "Verify login with valid username and password",
    "Verify login with invalid username",
    "Verify login with invalid password",
    "Verify login with blank username and password",
    "Verify error message is displayed for invalid credentials",
]

print("Requirement:")
print(requirement)

print("\nGenerated Test Cases:")
for index, test_case in enumerate(test_cases, start=1):
    print(f"{index}. {test_case}")
````
